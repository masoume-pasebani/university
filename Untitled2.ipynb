```python
# Import necessary libraries again after restarting the runtime
import whisper
from pydub import AudioSegment
from google.colab import files
from IPython.display import Audio, display
from hazm import Normalizer

# Load the Whisper model
model = whisper.load_model("medium")  # Try "large" for better results

def transcribe_audio(file_path):
    """
    Transcribes the audio file and normalizes Persian text if detected.
    """
    print("Transcribing... please wait ⏳")

    # Transcribe with language hint (Persian)
    result = model.transcribe(file_path, language='fa')  # Persian language hint
    lang = result['language']
    text = result['text']

    print(f"Detected language: {lang}")

    if lang == 'fa':
        print("Normalizing Persian text...")
        normalizer = Normalizer()
        text = normalizer.normalize(text)

    # Save the transcription to a text file
    output_file = "transcription.txt"
    with open(output_file, "w", encoding="utf-8") as f:
        f.write(text)

    print("✅ Transcription complete. You can download the file below.")
    files.download(output_file)

    return text

def upload_and_transcribe():
    """
    Facilitates the upload of an audio file and initiates the transcription process.
    """
    print("📂 Please upload an audio file:")
    uploaded = files.upload()

    for filename in uploaded.keys():
        print(f"📁 File {filename} uploaded successfully.")
        display(Audio(filename, autoplay=True))

        # Perform transcription with the larger model
        transcription = transcribe_audio(filename)

        print("\n📝 Transcribed Text:\n")
        print(transcription)

# Run the upload and transcribe function
upload_and_transcribe()
```
